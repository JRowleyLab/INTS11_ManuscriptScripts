```python
#filename="mergedloops_raw.txt"
filename="masterloops2"
ploc1 = 0
ploc2 = 0
oldscore = 0
outfile = open("master_mergedloops2.txt","w")
with open(filename) as fn:
    for line in fn:
        line2 = line.strip('\n')
        re = line2.split('\t')
        #print(re)
        newloc1 = int(re[1])
        newloc2 = int(re[2])
        newscore = re[3]
        d1 = abs(newloc1 - ploc1)
        d2 = abs(newloc2 - ploc2)
        #print(d1)
        if d1 <= 20000 and d2 <= 20000:
            if newscore > oldscore:
                outfile.write(str(re[0]) + ":" + str(newloc1) + ":" + str(newloc2) + "\n")
                
            else:
                outfile.write(str(re[0]) + ":" + str(ploc1) + ":" + str(ploc2) + "\n")
        else:
            outfile.write(str(re[0]) + ":" + str(newloc1) + ":" + str(newloc2) + "\n")
        ploc1 = newloc1
        ploc2 = newloc2
        oldscore = newscore
        
```
