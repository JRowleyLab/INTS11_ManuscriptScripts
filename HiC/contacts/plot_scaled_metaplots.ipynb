```python
import numpy as np
import seaborn as sns
```

```python
mat1 = np.loadtxt("genes_ctl_intragenic_1kbave.txt")
mat2 = np.loadtxt("genes_aux_intragenic_1kbave.txt")
mat3 = np.loadtxt("genes_noaux_intragenic_1kbave.txt")
mymin=0
mymax=1.3
```

```python
sns.heatmap(mat1[:41,:41], square=True, vmin=mymin,vmax=mymax)
```

```python
sns.heatmap(mat2[:41,:41], square=True, vmin=mymin,vmax=mymax)
```

```python
sns.heatmap(mat3[:41,:41], square=True, vmin=mymin,vmax=mymax)
```

```python
sns.heatmap((mat2[:41,:41]-mat1[:41,:41]), square=True, cmap="RdBu_r", vmin=-.2, vmax=.2)
```

```python
sns.heatmap((mat2[:41,:41]-mat3[:41,:41]), square=True, cmap="RdBu_r", vmin=-.2, vmax=.2)
```

```python
print(np.mean(mat1[20:21,20:21]))
print(np.mean(mat2[20:21,20:21]))
print(np.mean(mat3[20:21,20:21]))
```

```python
print(np.mean(mat1[23:,23:]))
print(np.mean(mat2[23:,23:]))
print(np.mean(mat3[23:,23:]))
```
